```python
# 1. Total Revenue

gross_revenue = ecommerce_data["revenue"].sum()

print("Total Revenue:", gross_revenue)

# gross_revenue → total revenue store panna variable.
# ecommerce_data["revenue"] → revenue column-a edukkum.
# .sum() → ellaa revenue values-um add pannum.
# print() → result screen-la show pannum.
```

```python
# 2. Total Quantity Sold

units_sold = ecommerce_data["quantity"].sum()

print("Total Units Sold:", units_sold)
# units_sold → total products quantity.
# quantity → ethana products sell aachu.
# .sum() → total quantity add pannum.
```

```python
# 3. Average Customer Rating

customer_score = ecommerce_data["customer_rating"].mean()
customer_score = round(customer_score,2)
print("Average Customer Rating:", customer_score)
# customer_score → average rating store pannum.
# .mean() → average calculate pannum.
# round() = A number-ah required decimal places-ku round off panna use panrom.
```

```python
# 4. Category-wise Revenue 

category_performance = ecommerce_data.groupby(
    "product_category"
)["revenue"].sum()

print(category_performance)

# groupby() → same category values-a group pannum.

```

```python

payment_category = ecommerce_data.pivot_table(
    values="revenue",                             #values → Revenue
    index="payment_method",                        #index → Payment
     columns="product_category",                   # columns → Category
    aggfunc="sum"                                  # aggfunc → Total
  )                                           
    
plt.figure(figsize=(10, 6))

sns.heatmap(
    payment_category,
    annot=True,
    fmt=".0f"
)

plt.title("Revenue by Payment Method and Product Category")
plt.xlabel("Product Category")
plt.ylabel("Payment Method")

plt.tight_layout()
plt.show()
# annot = show value
# fmt = format value  ,  fmt=".2f"  → 12543.78 & fmt=".0f"  → 12544
# Create a table by calculating the total revenue based on payment method and product category.
```

```python

# Create 4-in-1 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# -----------------------------------------
# 1. AREA CHART - Daily Revenue
# -----------------------------------------

daily_revenue = ecommerce_data.groupby(
    "order_date"
)["revenue"].sum()

axes[0, 0].fill_between(
    daily_revenue.index,
    daily_revenue.values
)

axes[0, 0].set_title("Daily Revenue Trend")
axes[0, 0].set_xlabel("Order Date")
axes[0, 0].set_ylabel("Revenue")


# -----------------------------------------
# 2. STEM PLOT - Quantity by Category
# -----------------------------------------

category_quantity = ecommerce_data.groupby(
    "product_category"
)["quantity"].sum()

axes[0, 1].stem(
    category_quantity.index,
    category_quantity.values
)

axes[0, 1].set_title("Quantity Sold by Product Category")
axes[0, 1].set_xlabel("Product Category")
axes[0, 1].set_ylabel("Quantity Sold")

axes[0, 1].tick_params(axis="x", rotation=45)


# -----------------------------------------
# 3. VIOLIN PLOT - Revenue by Payment Method
# -----------------------------------------

payment_methods = ecommerce_data[
    "payment_method"
].unique()

revenue_groups = []

for method in payment_methods:
    revenue_groups.append(
        ecommerce_data[
            ecommerce_data["payment_method"] == method
        ]["revenue"]
    )

axes[1, 0].violinplot(
    revenue_groups,
    showmedians=True
)

axes[1, 0].set_title(
    "Revenue Distribution by Payment Method"
)

axes[1, 0].set_xlabel("Payment Method")
axes[1, 0].set_ylabel("Revenue")

axes[1, 0].set_xticks(
    range(1, len(payment_methods) + 1)
)

axes[1, 0].set_xticklabels(
    payment_methods
)


# -----------------------------------------
# 4. HORIZONTAL BAR - Delivery Days
# -----------------------------------------

region_delivery = ecommerce_data.groupby(
    "region"
)["delivery_days"].mean().sort_values()

axes[1, 1].barh(
    region_delivery.index,
    region_delivery.values
)

axes[1, 1].set_title(
    "Average Delivery Days by Region"
)

axes[1, 1].set_xlabel("Average Delivery Days")
axes[1, 1].set_ylabel("Region")


# -----------------------------------------
# MAIN DASHBOARD TITLE
# -----------------------------------------

fig.suptitle(
    "E-COMMERCE SALES PERFORMANCE DASHBOARD",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
# 🔹 Bubble Chart — Quantity vs Unit Price, Revenue size
# 🔹 Step Chart — Cumulative Revenue
# 🔹 Error Bar Chart — Average Customer Rating by Region
# 🔹 Donut Chart — Orders by Payment Method

# Product Quantity vs Average Unit Price
# Cumulative Revenue Growth
# Average Customer Rating by Region
# Order Distribution by Payment Method

# Create 4-in-1 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. BUBBLE CHART ==> Bubble Chart — Quantity vs Unit Price, Revenue size


bubble_data = ecommerce_data.groupby(
    "product_category"
).agg({
    "quantity": "sum",
    "unit_price": "mean",
    "revenue": "sum"
})

axes[0, 0].scatter(
    bubble_data["quantity"],
    bubble_data["unit_price"],
    s=bubble_data["revenue"] / 5,
    alpha=0.6
)

axes[0, 0].set_title(
    "Product Quantity vs Average Unit Price"
)
axes[0, 0].set_xlabel("Total Quantity")
axes[0, 0].set_ylabel("Average Unit Price")


# 2. STEP CHART ==>  Step Chart — Cumulative Revenue

daily_revenue = ecommerce_data.groupby(
    "order_date"
)["revenue"].sum()

cumulative_revenue = daily_revenue.cumsum()

axes[0, 1].step(
    cumulative_revenue.index,
    cumulative_revenue.values,
    where="mid"
)

axes[0, 1].set_title(
    "Cumulative Revenue Growth"
)
axes[0, 1].set_xlabel("Order Date")
axes[0, 1].set_ylabel("Cumulative Revenue")



# 3. ERROR BAR CHART ==> Error Bar Chart — Average Customer Rating by Region

rating_by_region = ecommerce_data.groupby(
    "region"
)["customer_rating"].mean()

rating_std = ecommerce_data.groupby(
    "region"
)["customer_rating"].std()

axes[1, 0].errorbar(
    rating_by_region.index,
    rating_by_region.values,
    yerr=rating_std.values,
    fmt="o",
    capsize=5
)

axes[1, 0].set_title(
    "Average Customer Rating by Region"
)
axes[1, 0].set_xlabel("Region")
axes[1, 0].set_ylabel("Customer Rating")

# 4. DONUT CHART ==> Donut Chart — Orders by Payment Method

payment_orders = ecommerce_data[
    "payment_method"
].value_counts()

axes[1, 1].pie(
    payment_orders.values,
    labels=payment_orders.index,
    autopct="%1.1f%%",
    startangle=90
)

centre_circle = plt.Circle(
    (0, 0),
    0.70,
    fc="white"
)

axes[1, 1].add_artist(
    centre_circle
)

axes[1, 1].set_title(
    "Order Distribution by Payment Method"
)

# DASHBOARD TITLE

fig.suptitle(
    "E-COMMERCE SALES ANALYTICS DASHBOARD",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

# -----------------------------------------
# LOAD DATA
# -----------------------------------------

ecommerce_data = pd.read_csv(
    "ecommerce_sales_analytics_5000.csv"
)

# Convert date
ecommerce_data["order_date"] = pd.to_datetime(
    ecommerce_data["order_date"]
)


# -----------------------------------------
# DASHBOARD LAYOUT
# -----------------------------------------

fig = plt.figure(figsize=(18, 14))

grid = GridSpec(
    3,
    3,
    figure=fig,
    width_ratios=[1.2, 1.2, 0.8],
    height_ratios=[1.2, 1, 1],
    wspace=0.35,
    hspace=0.45
)


# =================================================
# 1. MONTHLY REVENUE TREND
# BIG CHART
# =================================================

monthly_revenue = ecommerce_data.groupby(
    ecommerce_data["order_date"].dt.to_period("M")
)["revenue"].sum()

ax1 = fig.add_subplot(
    grid[0, :2]
)

ax1.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o"
)

ax1.set_title(
    "Monthly Revenue Trend"
)

ax1.set_xlabel(
    "Month"
)

ax1.set_ylabel(
    "Revenue"
)

ax1.tick_params(
    axis="x",
    rotation=45
)


# =================================================
# 2. PAYMENT METHOD DISTRIBUTION
# SMALL CHART
# =================================================

payment_orders = ecommerce_data[
    "payment_method"
].value_counts()

ax2 = fig.add_subplot(
    grid[0, 2]
)

ax2.pie(
    payment_orders.values,
    labels=payment_orders.index,
    autopct="%1.1f%%",
    startangle=90
)

ax2.set_title(
    "Payment Method Mix"
)


# =================================================
# 3. CATEGORY PERFORMANCE
# MEDIUM CHART
# =================================================

category_data = ecommerce_data.groupby(
    "product_category"
)["revenue"].sum().sort_values()

ax3 = fig.add_subplot(
    grid[1, 0]
)

ax3.barh(
    category_data.index,
    category_data.values
)

ax3.set_title(
    "Category Revenue Performance"
)

ax3.set_xlabel(
    "Revenue"
)


# =================================================
# 4. QUANTITY vs REVENUE
# BIG BUBBLE CHART
# =================================================

product_analysis = ecommerce_data.groupby(
    "product_category"
).agg({
    "quantity": "sum",
    "revenue": "sum",
    "unit_price": "mean"
})

ax4 = fig.add_subplot(
    grid[1, 1:]
)

ax4.scatter(
    product_analysis["quantity"],
    product_analysis["revenue"],
    s=product_analysis["unit_price"] * 8,
    alpha=0.6
)

for category in product_analysis.index:

    ax4.annotate(
        category,
        (
            product_analysis.loc[
                category,
                "quantity"
            ],
            product_analysis.loc[
                category,
                "revenue"
            ]
        )
    )

ax4.set_title(
    "Quantity vs Revenue by Category"
)

ax4.set_xlabel(
    "Total Quantity Sold"
)

ax4.set_ylabel(
    "Total Revenue"
)


# =================================================
# 5. REGIONAL REVENUE HEATMAP
# BIG CHART
# =================================================

region_category = pd.crosstab(
    ecommerce_data["region"],
    ecommerce_data["product_category"],
    values=ecommerce_data["revenue"],
    aggfunc="sum"
)

ax5 = fig.add_subplot(
    grid[2, :2]
)

image = ax5.imshow(
    region_category.values,
    aspect="auto"
)

ax5.set_title(
    "Regional Revenue by Product Category"
)

ax5.set_xlabel(
    "Product Category"
)

ax5.set_ylabel(
    "Region"
)

ax5.set_xticks(
    range(len(region_category.columns))
)

ax5.set_xticklabels(
    region_category.columns,
    rotation=45
)

ax5.set_yticks(
    range(len(region_category.index))
)

ax5.set_yticklabels(
    region_category.index
)

fig.colorbar(
    image,
    ax=ax5,
    label="Revenue"
)


# =================================================
# 6. DELIVERY PERFORMANCE
# MEDIUM CHART
# =================================================

delivery_analysis = ecommerce_data.groupby(
    "delivery_days"
)["customer_rating"].mean()

ax6 = fig.add_subplot(
    grid[2, 2]
)

ax6.plot(
    delivery_analysis.index,
    delivery_analysis.values,
    marker="o"
)

ax6.set_title(
    "Delivery Days vs Rating"
)

ax6.set_xlabel(
    "Delivery Days"
)

ax6.set_ylabel(
    "Average Rating"
)


# =================================================
# MAIN DASHBOARD TITLE
# =================================================

fig.suptitle(
    "E-COMMERCE SALES ANALYTICS DASHBOARD",
    fontsize=20
)

plt.tight_layout(
    rect=[0, 0, 1, 0.96]
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.patches import FancyBboxPatch
import numpy as np

# --------------------------------------------------
# 1. LOAD DATA
# --------------------------------------------------

ecommerce_data = pd.read_csv("ecommerce_sales_analytics_5000.csv")

ecommerce_data["order_date"] = pd.to_datetime(
    ecommerce_data["order_date"]
)

# --------------------------------------------------
# 2. KPI VALUES
# --------------------------------------------------

total_revenue = ecommerce_data["revenue"].sum()
total_orders = ecommerce_data["order_id"].nunique()
total_quantity = ecommerce_data["quantity"].sum()
average_rating = ecommerce_data["customer_rating"].mean()

# --------------------------------------------------
# 3. ANALYSIS
# --------------------------------------------------

monthly_revenue = ecommerce_data.groupby(
    ecommerce_data["order_date"].dt.month
)["revenue"].sum()

category_revenue = ecommerce_data.groupby(
    "product_category"
)["revenue"].sum().sort_values()

region_revenue = ecommerce_data.groupby(
    "region"
)["revenue"].sum()

payment_orders = ecommerce_data[
    "payment_method"
].value_counts()

rating_by_category = ecommerce_data.groupby(
    "product_category"
)["customer_rating"].mean()

delivery_by_region = ecommerce_data.groupby(
    "region"
)["delivery_days"].mean()

# --------------------------------------------------
# 4. CREATE DASHBOARD
# --------------------------------------------------

fig = plt.figure(figsize=(20, 14))

fig.patch.set_facecolor("#111827")

grid = GridSpec(
    5,
    4,
    figure=fig,
    height_ratios=[0.8, 1.5, 2.5, 2.5, 1.2],
    hspace=0.8,
    wspace=0.5
)

# --------------------------------------------------
# 5. DASHBOARD TITLE
# --------------------------------------------------

fig.text(
    0.05,
    0.965,
    "E-COMMERCE SALES PERFORMANCE",
    fontsize=24,
    fontweight="bold",
    color="white"
)

fig.text(
    0.05,
    0.935,
    "Business Intelligence | Executive Analytics Dashboard",
    fontsize=11,
    color="lightgray"
)

# --------------------------------------------------
# 6. KPI CARDS
# --------------------------------------------------

kpi_values = [
    f"₹{total_revenue:,.0f}",
    f"{total_orders:,}",
    f"{total_quantity:,}",
    f"{average_rating:.2f}"
]

kpi_titles = [
    "TOTAL REVENUE",
    "TOTAL ORDERS",
    "UNITS SOLD",
    "AVG CUSTOMER RATING"
]

for i in range(4):

    ax = fig.add_subplot(grid[1, i])

    ax.set_facecolor("#1f2937")

    card = FancyBboxPatch(
        (0, 0),
        1,
        1,
        boxstyle="round,pad=0.03,rounding_size=0.05",
        transform=ax.transAxes,
        linewidth=1.5,
        edgecolor="#374151",
        facecolor="#1f2937"
    )

    ax.add_patch(card)

    ax.text(
        0.5,
        0.60,
        kpi_values[i],
        ha="center",
        va="center",
        fontsize=22,
        fontweight="bold",
        color="white"
    )

    ax.text(
        0.5,
        0.25,
        kpi_titles[i],
        ha="center",
        va="center",
        fontsize=9,
        color="lightgray"
    )

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

# --------------------------------------------------
# 7. MONTHLY REVENUE TREND
# --------------------------------------------------

ax1 = fig.add_subplot(grid[2, :2])

ax1.set_facecolor("#1f2937")

ax1.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o",
    linewidth=3
)

ax1.fill_between(
    monthly_revenue.index,
    monthly_revenue.values,
    alpha=0.15
)

ax1.set_title(
    "Revenue Performance Trend",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax1.set_xlabel(
    "Month",
    color="lightgray"
)

ax1.set_ylabel(
    "Revenue",
    color="lightgray"
)

ax1.tick_params(
    colors="lightgray"
)

for spine in ax1.spines.values():
    spine.set_visible(False)

# --------------------------------------------------
# 8. CATEGORY PERFORMANCE
# --------------------------------------------------

ax2 = fig.add_subplot(grid[2, 2:])

ax2.set_facecolor("#1f2937")

ax2.stem(
    category_revenue.index,
    category_revenue.values
)

ax2.set_title(
    "Product Category Revenue",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax2.set_xlabel(
    "Product Category",
    color="lightgray"
)

ax2.set_ylabel(
    "Revenue",
    color="lightgray"
)

ax2.tick_params(
    axis="x",
    rotation=45,
    colors="lightgray"
)

ax2.tick_params(
    axis="y",
    colors="lightgray"
)

for spine in ax2.spines.values():
    spine.set_visible(False)

# --------------------------------------------------
# 9. REGIONAL REVENUE
# --------------------------------------------------

ax3 = fig.add_subplot(grid[3, :2])

ax3.set_facecolor("#1f2937")

angles = np.linspace(
    0,
    2 * np.pi,
    len(region_revenue),
    endpoint=False
)

values = region_revenue.values

ax3 = plt.subplot(
    grid[3, :2],
    projection="polar"
)

ax3.set_facecolor("#1f2937")

ax3.bar(
    angles,
    values,
    width=0.6,
    alpha=0.8
)

ax3.set_title(
    "Regional Revenue Performance",
    color="white",
    fontsize=14,
    fontweight="bold",
    pad=20
)

ax3.set_xticks(angles)

ax3.set_xticklabels(
    region_revenue.index,
    color="white"
)

ax3.set_yticklabels([])

# --------------------------------------------------
# 10. PAYMENT BEHAVIOUR
# --------------------------------------------------

ax4 = fig.add_subplot(grid[3, 2])

ax4.set_facecolor("#1f2937")

ax4.pie(
    payment_orders.values,
    labels=payment_orders.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.45)
)

ax4.set_title(
    "Payment Method Usage",
    color="white",
    fontsize=14,
    fontweight="bold"
)

# --------------------------------------------------
# 11. CUSTOMER EXPERIENCE
# --------------------------------------------------

ax5 = fig.add_subplot(grid[3, 3])

ax5.set_facecolor("#1f2937")

ax5.bar(
    rating_by_category.index,
    rating_by_category.values
)

ax5.set_title(
    "Average Rating by Category",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax5.set_xlabel(
    "Category",
    color="lightgray"
)

ax5.set_ylabel(
    "Rating",
    color="lightgray"
)

ax5.tick_params(
    axis="x",
    rotation=45,
    colors="lightgray"
)

ax5.tick_params(
    axis="y",
    colors="lightgray"
)

for spine in ax5.spines.values():
    spine.set_visible(False)

# --------------------------------------------------
# 12. EXECUTIVE INSIGHTS
# --------------------------------------------------

best_category = category_revenue.idxmax()
best_region = region_revenue.idxmax()
best_payment = payment_orders.idxmax()
average_delivery = ecommerce_data["delivery_days"].mean()

insight_text = (
    "EXECUTIVE BUSINESS INSIGHTS\n\n"
    f"• Best Product Category: {best_category}\n"
    f"• Highest Revenue Region: {best_region}\n"
    f"• Most Used Payment Method: {best_payment}\n"
    f"• Average Delivery Time: {average_delivery:.2f} days\n"
    f"• Average Customer Rating: {average_rating:.2f}/5"
)

ax6 = fig.add_subplot(grid[4, :])

ax6.set_facecolor("#1f2937")

ax6.text(
    0.02,
    0.5,
    insight_text,
    fontsize=12,
    color="white",
    verticalalignment="center"
)

ax6.axis("off")

# --------------------------------------------------
# 13. FINAL DASHBOARD
# --------------------------------------------------

plt.show()
```
